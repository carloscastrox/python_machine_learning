- Correlación cercana a 0 → no existe relación fuerte.

Las métricas permiten medir qué tan bueno es el modelo.

| Métrica | Nombre en Español            | ¿Qué mide?                                                                                    | Interpretación                                                |
| ------- | ---------------------------- | --------------------------------------------------------------------------------------------- | ------------------------------------------------------------- |
| MAE     | Error Absoluto Medio         | Calcula el error promedio entre las predicciones y los valores reales.                        | Mientras más cercano a 0, mejor es el modelo.                 |
| MSE     | Error Cuadrático Medio       | Calcula el promedio de los errores elevados al cuadrado, penalizando más los errores grandes. | Valores pequeños indican menos errores graves.                |
| R²      | Coeficiente de Determinación | Mide qué tan bien el modelo logra explicar el comportamiento de los datos.                    | 1 = excelente, 0 = muy malo. Mientras más cercano a 1, mejor. |
